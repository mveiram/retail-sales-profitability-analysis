# Retail Sales & Profitability Analysis
## Notebook 01 — Exploratory Data Analysis (EDA)

**Objective:** understand the structure, quality and initial business patterns in the Superstore dataset before building the SQL analysis and Power BI dashboard.

**Dataset:** Sample - Superstore  
**Tools:** Python, Pandas, Matplotlib

## 1. Business Questions

1. How are sales and profit distributed?
2. Which categories and regions contribute most to performance?
3. Does discounting appear to be associated with lower profitability?
4. Which products generate negative profit?
5. What findings should influence the Power BI dashboard?

## 2. Load the Data

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/Sample - Superstore.csv', encoding='latin1')
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

print(f'Rows: {df.shape[0]:,}')
print(f'Columns: {df.shape[1]}')
```

## 3. Data Quality

The source dataset contains **9,994 rows and 21 columns**. Initial validation found:

- **0 duplicate rows**
- **0 missing values**
- Date fields converted to datetime
- Numeric fields available for Sales, Quantity, Discount and Profit

No broad missing-value treatment is required at this stage. Any later transformation should be justified by the analytical objective.

```python
display(df.head())
display(df.dtypes.to_frame('dtype'))

quality = pd.DataFrame({
    'null_count': df.isna().sum(),
    'null_%': (df.isna().mean() * 100).round(2),
    'unique_values': df.nunique()
})
display(quality.sort_values('null_count', ascending=False))

print('Duplicate rows:', df.duplicated().sum())
```

## 4. Descriptive Statistics

```python
display(df[['Sales', 'Quantity', 'Discount', 'Profit']].describe().T)
```

## 5. Core Business KPIs

| KPI | Value |
|---|---:|
| Total Sales | $2,297,200.86 |
| Total Profit | $286,397.02 |
| Orders | 5,009 |
| Customers | 793 |
| Profit Margin | 12.47% |
| Average Order Value | $458.61 |

These KPIs will become the foundation of the Power BI Executive Overview.

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()
profit_margin = total_profit / total_sales
avg_order_value = total_sales / total_orders

kpis = pd.Series({
    'Total Sales': total_sales,
    'Total Profit': total_profit,
    'Orders': total_orders,
    'Customers': total_customers,
    'Profit Margin': profit_margin,
    'Average Order Value': avg_order_value
})
display(kpis.to_frame('value'))
```

## 6. Category Performance

Initial findings:

- **Technology** generates the highest sales and has a strong profit margin.
- **Office Supplies** also shows healthy profitability.
- **Furniture** generates substantial sales but has a much lower profit margin than the other two categories.

Furniture is therefore an important area for deeper profitability analysis, particularly at the sub-category and discount levels.

```python
category = (df.groupby('Category', as_index=False)
             .agg(Sales=('Sales','sum'), Profit=('Profit','sum')))
category['Profit Margin'] = category['Profit'] / category['Sales']
display(category.sort_values('Sales', ascending=False))
```

```python
plt.figure(figsize=(8,5))
plt.bar(category['Category'], category['Sales'])
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

## 7. Regional Performance

Initial findings:

- The **West** region has the highest profit margin among the four regions.
- The **Central** region has the lowest profit margin.
- East generates substantial sales and profit, while South has the lowest sales volume.

Regional analysis should therefore compare **sales and profitability**, rather than using sales alone as the performance indicator.

```python
region = (df.groupby('Region', as_index=False)
           .agg(Sales=('Sales','sum'), Profit=('Profit','sum')))
region['Profit Margin'] = region['Profit'] / region['Sales']
display(region.sort_values('Sales', ascending=False))
```

## 8. Discount vs. Profitability

The next step is to test whether higher discount levels are associated with weaker margins. This is an important business question because discounting can increase sales volume while reducing profitability.

```python
discount_profit = (df.groupby('Discount', as_index=False)
                    .agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Orders=('Order ID','nunique')))
discount_profit['Profit Margin'] = discount_profit['Profit'] / discount_profit['Sales']
display(discount_profit.sort_values('Discount'))
```

## 9. Loss-Making Products

The initial product-level analysis identifies products with negative total profit. These products should be investigated further using discount, category, quantity and regional dimensions.

```python
loss_products = (
    df.groupby(['Product ID', 'Product Name'])
      .agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Quantity=('Quantity','sum'))
      .sort_values('Profit').head(10).reset_index()
)
display(loss_products)
```

## 10. Initial Business Insights

1. Overall profitability is positive, with a **12.47% profit margin**.
2. **Furniture requires deeper investigation** because its profit margin is materially lower than Technology and Office Supplies.
3. **Regional performance differs significantly**, making geography relevant for the dashboard.
4. Negative-profit products exist despite generating meaningful sales, so revenue alone is not an adequate performance measure.
5. Discount level should be tested as a potential driver of margin differences.

These are **initial observations**, not final recommendations. The SQL analysis will validate and extend them before the Power BI dashboard is finalized.
